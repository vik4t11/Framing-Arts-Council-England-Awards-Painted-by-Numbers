- Load libraries and datasets

- Initial datasets review 

- Standardise column names 

- Ammend DataTypes 

- Identify missing, unique or duplicate values

- Add year columns for year-on-year analysis

- Join 2023-2024 and 2025-2025 datasets

- Drop identifiable information columns  

- Save DataFrame into clean data folder
