## <u>**Standardise Column Names**</u>

Format must be consistent to access columns in the EDA, Feature Engineering and ML Modeling notebooks. Remove leading or trailing whitespace use <code>.strip()</code>, replace spaces or hyphens with underscores use <code>.replace()</code> and convert text to lowercase use <code>.lower()</code>. These functions are a professional standard protocol for technical users to navigate the code cells.

### **Handle Missing Values**

Replace missing value in the 2025-2026 DataFrame for <code>local_authority</code> with "Unknown", as this recipient potentially operates in more than one local authority according to the Arts Council documentation. 

```python
# Replace missing value in the 'local_authority' column with 'Unknown'
df_26['local_authority'] = df_26['local_authority'].fillna('Unknown')

# Check for remaining missing values in df_26
df_26.isna().sum()
```

### **Award Amount**

Similar to standardising column names in the previous section <code>.replace()</code> and <code>.strip()</code> remove symbols and hidden spaces to avoid ValueError caused by an invalid literal. Convert string into numeric DataType use <code>.astype(int)</code>.

```python
# Clean and convert the 'award_amount' column in df_24
df_24["award_amount"] = (df_24["award_amount"]
                         .str.replace(',','')
                         .str.replace('£','')
                         .str.strip()
                         .astype(int))


# Clean and convert the 'award_amount' column in df_26
df_26["award_amount"] = (df_26["award_amount"]
                         .str.replace(',','')
                         .str.replace('£','')
                         .str.strip()
                         .astype(int))


# Check the data type of the 'award_amount' column
print("Data type df_24:", df_24["award_amount"].dtype) 
print("Data type df_26:", df_26["award_amount"].dtype) 
```

### **Date-Time**

Convert award-date column from string into date-time. Use <code>pd.to_datetime</code> for both DateFrames and check the conversion is successful.  

```python
# Convert award_date columns to datetime so .dt.year works
df_24['award_date'] = pd.to_datetime(df_24['award_date'], errors='coerce')
df_26['award_date'] = pd.to_datetime(df_26['award_date'], errors='coerce')
df_24['award_year'] = df_24['award_date'].dt.year

# Extract the year from the award_date column in df_26
df_26['award_year'] = df_26['award_date'].dt.year
```

```python
# Check the first few rows of df_24 to verify the changes
df_24.head()   
```

Add a column dedicated to year and improve filtering or grouping rows. Use <code>.dt.year</code> to extract the year from award date column.  

```python
# Remove time from award_date column in df_24 and df_26
df_24['award_date'] = df_24['award_date'].dt.date
df_26['award_date'] = df_26['award_date'].dt.date
```

### **Categorical**

Categorical data types contain a list of unique values that represent different categories. Use <code>.astype("category")</code> to convert strings into a categorical data type. 

```python
# Group categorical columns for conversion to 'category' data type
categorical_cols = [
    "main_discipline",
    "strand",
    "time_limited_priority",
    "ace_area"
]

# Convert the categorical columns
for df in [df_24, df_26]:
    for col in categorical_cols:
        if col in df.columns:
            df[col] = df[col].astype("category")
```
